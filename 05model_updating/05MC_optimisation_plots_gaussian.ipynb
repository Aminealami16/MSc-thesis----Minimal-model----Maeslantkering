## **2D plots (histograms and heat maps)**

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes, mark_inset

success_grid = np.full((N2, N2), np.nan)

for _, row in data_42.iterrows():
    success_grid[int(row["i"]), int(row["j"])] = row["success_rate"]

valid_values = success_grid[~np.isnan(success_grid)]

mean_val = np.mean(valid_values)
median_val = np.median(valid_values)
std_val = np.std(valid_values)

# SAME FIGURE SIZE AS YOUR BASE CASE
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# -------------------
# LEFT: HEATMAP
# -------------------
im = axes[0].imshow(
    success_grid,
    origin="lower",
    aspect="auto",
    extent=[kf_min2, kf_max2, Iy_min2, Iy_max2],
    cmap="viridis"
)

axes[0].set_xlabel(r"$k_f$ [N/m]")
axes[0].set_ylabel(r"$I_y$ [m⁴]")

# FIXED COLORBAR (DO NOT USE make_axes_locatable)
cbar = fig.colorbar(im, ax=axes[0], fraction=0.046, pad=0.04)
cbar.set_label("Success rate")

# -------------------
# DATA FOR INSET
# -------------------
rows, cols = np.where(~np.isnan(success_grid))

x_vals = np.linspace(kf_min2, kf_max2, N2)[cols]
y_vals = np.linspace(Iy_min2, Iy_max2, N2)[rows]
z_vals = success_grid[~np.isnan(success_grid)]

x_zoom_min, x_zoom_max = np.min(x_vals), np.max(x_vals)
y_zoom_min, y_zoom_max = np.min(y_vals), np.max(y_vals)

x_pad = 0.05 * (x_zoom_max - x_zoom_min)
y_pad = 0.05 * (y_zoom_max - y_zoom_min)

x_zoom_min -= x_pad
x_zoom_max += x_pad
y_zoom_min -= y_pad
y_zoom_max += y_pad

rect = plt.Rectangle(
    (x_zoom_min, y_zoom_min),
    x_zoom_max - x_zoom_min,
    y_zoom_max - y_zoom_min,
    linewidth=1.5,
    edgecolor="black",
    facecolor="none"
)
axes[0].add_patch(rect)

# INSET
axins = inset_axes(
    axes[0],
    width="62%",
    height="62%",
    loc="upper right",
    borderpad=0.8
)

axins.scatter(
    x_vals,
    y_vals,
    c=z_vals,
    cmap="viridis",
    s=10,
    edgecolors="none"
)

axins.set_xlim(x_zoom_min, x_zoom_max)
axins.set_ylim(y_zoom_min, y_zoom_max)
axins.tick_params(labelsize=8)

mark_inset(axes[0], axins, loc1=2, loc2=4, fc="none", ec="black", lw=1)

# -------------------
# RIGHT: HISTOGRAM
# -------------------
axes[1].hist(
    valid_values,
    bins="fd",
    color="steelblue",
    edgecolor="black",
    alpha=0.85
)

axes[1].set_xlabel("Success rate")
axes[1].set_ylabel("Frequency")
axes[1].set_title("Distribution of success rates (4 modes & 2 parameters)")
axes[1].set_xlim(0, 1)

textstr = (
    f"Mean:   {mean_val:.3f}\n"
    f"Median: {median_val:.3f}\n"
    f"Std:    {std_val:.3f}"
)

axes[1].text(
    0.95, 0.95,
    textstr,
    transform=axes[1].transAxes,
    fontsize=10,
    va="top",
    ha="right",
    bbox=dict(boxstyle="round", facecolor="white", edgecolor="gray", alpha=0.9)
)

# -------------------
# CONSISTENT LAYOUT FOR THESIS STACKING
# -------------------
fig.subplots_adjust(left=0.08, right=0.95, wspace=0.35)

fig.patch.set_facecolor("white")
axes[0].set_facecolor("white")
axes[1].set_facecolor("white")

# IMPORTANT: do NOT crop
plt.savefig(
    "csv_files/success_rate_distribution_2D_42_smart.pdf",
    format="pdf",
    bbox_inches=None,
    facecolor="white"
)

plt.show()
```
